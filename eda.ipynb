```python
df["Order Date"].dt.year
```

```python
yearly_sales = df.groupby(df["Order Date"].dt.year)["Sales"].sum()
yearly_growth = yearly_sales.pct_change() * 100
yearly_growth
```

```python
yearly_sales = df.groupby(df["Order Date"].dt.year)["Sales"].sum()

yearly_sales.plot(kind="line", marker="o")

plt.title("Annual Sales Trend")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.xticks(yearly_sales.index)

plt.show()
```

```python
monthly_sales = df.groupby([
    df["Order Date"].dt.year,
    df["Order Date"].dt.month
])["Sales"].sum()
monthly_sales = monthly_sales.unstack()
monthly_sales
```

```python
monthly_sales.T.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend by Year")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(range(1, 13))

plt.show()
```

```python
customer_freq = df.groupby("Customer Name")["Order ID"].nunique().sort_values(ascending=False)
customer_freq.head(10)
```

```python
aov = df["Sales"].sum() / df["Order ID"].nunique()
aov

```

```python
top_products_sales = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

top_products_sales.head(10)
```

```python
top_products_profit = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

top_products_profit.head(10)
```

```python
worst_products_profit = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values()
)

worst_products_profit.head(10)
```

```python
subcategory_summary = df.groupby("Sub-Category").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_values("Sales", ascending=False)

subcategory_summary
```

```python
subcategory_summary.sort_values("Profit", ascending=False).head(10)
```

```python
subcategory_summary.sort_values("Profit").head(10)
```

```python
region_summary = df.groupby("Region").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_values("Sales", ascending=False)

region_summary
```

```python
region_summary["Profit Margin %"] = (
    region_summary["Profit"] / region_summary["Sales"] * 100
).round(2)

region_summary
```

```python
state_summary = df.groupby("State").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).sort_values("Sales", ascending=False)

state_summary.head(10)
```

```python
state_summary.sort_values("Profit").head(10)
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_products = df["Product ID"].nunique()
total_quantity = df["Quantity"].sum()

profit_margin = (total_profit / total_sales) * 100
aov = total_sales / total_orders

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Unique Products:", total_products)
print("Total Quantity:", total_quantity)
print("Profit Margin:", round(profit_margin, 2), "%")
print("Average Order Value:", round(aov, 2))
```
